**Fragility Analysis**


The 2 fragile routes:

1. by_name(`df[["company","seats","mrr"]]`): This depends on naming, so if a colleague renames any of these columns, the code will fail cause the index labels don't exist anymore

2. by `.iloc` positions(`df.iloc[:,[1,4,5]]`): This depends on column position, i.e if a colleague reorders(adds,removes or changes) columns, the positions may change and return the wrong columns

**Production Choice with Justification**

I would put `.loc` in a production code because it is clear, easy understand,selects all rows and does not depend on columns staying in order,changes in column order does not affect result.

##### *Q8: Parsing is not validating*

````python
#code that flags both kinds

#code for parsing date columns
df["signup_ts"] = pd.to_datetime(df["signup_ts"],errors="coerce")
df["last_active"] = pd.to_datetime(df["last_active"],errors="coerce")
df
````

````python
#dates that could not be converted(parsing problems)
parsing_errors = df[df["signup_ts"].isna() | df["last_active"].isna()]
parsing_errors
````

````python
#validation problems(valid but makes no sense)
#future_dates = df[(df["signup_ts"] > pd.Timestamp.now()) | (df["last_active"] > pd.Timestamp.now())]
#future_dates
before_signup = df[df["last_active"] < df["signup_ts"]]
before_signup
````

##**SUMMARY**
**Code used to identify both kinds of problems**

*Parse the date columns*

`df["signup_ts"] = pd.to_datetime(df["signup_ts"], errors="coerce")`

`df["last_active"] = pd.to_datetime(df["last_active"], errors="coerce")`


*Parsing problems (dates that could not be converted)*

parsing_errors = `df[df["signup_ts"].isna() | df["last_active"].isna()]`

*Validation problems (dates that are valid but do not make business sense)*

future_dates = `df[(df["signup_ts"] > pd.Timestamp.now()) | (df["last_active"] > pd.Timestamp.now())]`

before_signup = `df[df["last_active"] < df["signup_ts"]]`

**Parsing problem:** A parsing problem happens when Pandas cannot convert a value into a date. In this case, pd.to_datetime() returns NaT, showing that the value could not be interpreted as a valid date.

*For example:*
The signup_ts column contains different formats, including standard dates and a Unix timestamp (1675209600). If these formats are not handled correctly during parsing, some values may become NaT.

**Validation problem:** A validation problem occurs after the date is successfully converted, but it still does not make sense based on the business rules.

*For example:*
A1013 has last_active = 2099-01-01. Pandas accepts this as a valid date, but it is unrealistic because it is far in the future.

A1007 has signup_ts = 30/06/2023 but last_active = 15/06/2023. Both dates are valid, but a customer cannot be active before signing up.

These rows show why parsing alone is not enough. A date can have the correct format but still represent impossible business data.

**Difference between parsing and validation**

Parsing checks whether a value can be converted into a date.

Validation checks whether the date makes sense in the real world or according to the business rules.

A date can pass parsing but still fail validation, which is why both checks are necessary during data cleaning.
